## Manejar los valores faltantes en la columna 'cabin'

### Subtask:
Decide cómo imputar o eliminar los valores faltantes en la columna 'Cabin'.


**Reasoning**:
Verifica la cantidad de valores faltantes en la columna 'Cabin' para confirmar el alcance de los datos ausentes y luego decide si eliminar la columna basándote en esto.



```python
# Import pandas
import pandas as pd

# Import Kaggle
import kagglehub

# Reload the dataframe
path = kagglehub.dataset_download("brendan45774/test-file")
df = pd.read_csv(path + "/tested.csv")

# Verifica la cantidad de valores faltantes en la columna 'Cabin'
print("Valores faltantes por columna antes de manejar la columna Cabin:")
print(df.isnull().sum())

# Decide si eliminar la columna 'Cabin' debido a la gran cantidad de valores faltantes
# Como la columna 'Cabin' tiene 327 values faltantes de 418, es una proporción muy alta.
# Es mejor eliminar la columna para evitar introducir sesgos.
df.drop('Cabin', axis=1, inplace=True)

print("\nValores faltantes por columna después de eliminar la columna Cabin:")
print(df.isnull().sum())
```

## Transformación de variables categóricas

### Subtask:
Convierte las variables categóricas como 'Sex' y 'Embarked' a formatos numéricos si es necesario para el modelado.


**Reasoning**:
Convierte la columna 'Sex' a formato numérico usando codificación binaria y la columna 'Embarked' a formato numérico usando codificación one-hot, luego elimina las columnas categóricas originales.



```python
# Convert 'Sex' to numerical (0 for female, 1 for male)
df['Sex'] = df['Sex'].map({'female': 0, 'male': 1})

# Convert 'Embarked' to numerical using one-hot encoding
df = pd.get_dummies(df, columns=['Embarked'], prefix='Embarked')

# Display the first few rows to verify the transformations
display(df.head())
```

## Análisis univariado y bivariado

### Subtask:
Realiza análisis exploratorios adicionales, como la distribución de variables y la relación entre variables.


**Reasoning**:
Genera histogramas para variables numéricas y diagramas de barras para variables numéricas categóricas para visualizar sus distribuciones, y diagramas de dispersión o de cajas para las relaciones entre variables numéricas.



```python
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Histogramas para variables numéricas
numerical_cols = ['Age', 'Fare', 'SibSp', 'Parch']
df[numerical_cols].hist(bins=20, figsize=(12, 8))
plt.suptitle('Distribución de Variables Numéricas', y=1.02, fontsize=16)
plt.tight_layout()
plt.show()

# 2. Gráficos de barras para variables categóricas numéricas
categorical_num_cols = ['Survived', 'Pclass']
fig, axes = plt.subplots(1, len(categorical_num_cols), figsize=(10, 5))
for i, col in enumerate(categorical_num_cols):
    df[col].value_counts().sort_index().plot(kind='bar', ax=axes[i], color=sns.color_palette('viridis', n_colors=len(df[col].unique())))
    axes[i].set_title(f'Frecuencia de {col}')
    axes[i].set_xlabel(col)
    axes[i].set_ylabel('Frecuencia')
    axes[i].tick_params(axis='x', rotation=0)
plt.tight_layout()
plt.show()

# 3. Gráficos de dispersión para variables numéricas (Age vs Fare)
plt.figure(figsize=(8, 6))
sns.scatterplot(x='Age', y='Fare', data=df, alpha=0.6)
plt.title('Relación entre Edad y Tarifa')
plt.xlabel('Edad')
plt.ylabel('Tarifa')
plt.show()

# 4. Box plots para la relación entre variables categóricas y numéricas (Survived vs Age)
plt.figure(figsize=(8, 6))
sns.boxplot(x='Survived', y='Age', data=df)
plt.title('Distribución de Edad por Supervivencia')
plt.xlabel('Sobrevivió (0=No, 1=Sí)')
plt.ylabel('Edad')
plt.show()
```

## Preparación de datos para modelado

### Subtask:
Realiza las transformaciones finales necesarias en los datos para que estén listos para ser utilizados en modelos de Machine Learning.


# Transformación Final
Realiza las transformaciones finales de los datos, incluyendo el manejo de los valores faltantes restantes, la aplicación de codificación one-hot a cualquier característica categórica nominal que quede, considerando la normalización de las características numéricas si es necesario, y finalmente definiendo las características (X) y el objetivo (y) para el modelado de aprendizaje automático.

```python
# 1. Verifica si hay alguna columna que aún tenga valores faltantes
print("Valores faltantes por columna antes de las transformaciones finales:")
print(df.isnull().sum())

# Based on the previous output, 'Cabin' was dropped and 'Age' and 'Fare' were imputed.
# 'Name' and 'Ticket' are categorical but likely not useful features for modeling directly.
# We will drop 'Name' and 'Ticket' as they are not directly suitable for ML models without significant feature engineering (which is outside the scope of basic ETL).
df = df.drop(['Name', 'Ticket'], axis=1)

# 2. Si hay variables categóricas nominales que no han sido transformadas
# 'Sex' and 'Embarked' have already been transformed in previous steps.
# No other categorical nominal variables are identified as useful for direct modeling.

# 3. Considera si es necesario escalar las variables numéricas
# Scaling is often beneficial for models sensitive to the scale of data.
# Numerical columns are 'PassengerId', 'Age', 'SibSp', 'Parch', 'Fare'.
# 'PassengerId' is an identifier and should not be scaled or used as a feature.
# 'Survived', 'Pclass' are already in a suitable numerical format or are target/ordinal.
# We will scale 'Age', 'SibSp', 'Parch', and 'Fare'.
from sklearn.preprocessing import StandardScaler

numerical_features_to_scale = ['Age', 'SibSp', 'Parch', 'Fare']
scaler = StandardScaler()
df[numerical_features_to_scale] = scaler.fit_transform(df[numerical_features_to_scale])

print("\nDataFrame después de escalar las características numéricas:")
display(df.head())

# 4. Revisa los tipos de datos de todas las columnas
print("\nTipos de datos después de las transformaciones:")
print(df.info())

# 5. Define las variables predictoras (features, X) y la variable objetivo (target, y)
# 'Survived' is the target variable.
# All other columns except 'PassengerId' (which is just an identifier) will be features.
X = df.drop(['Survived', 'PassengerId'], axis=1)
y = df['Survived']

print("\nVariables predictoras (X):")
display(X.head())

print("\nVariable objetivo (y):")
display(y.head())
```

**Reasoning**:
El paso anterior mostró que 'Age' y 'Fare' todavía tienen valores faltantes antes de las transformaciones finales, a pesar de los intentos previos de imputarlos. La normalización se aplicó antes de imputar los NaN restantes en 'Age' y 'Fare', lo cual es incorrecto. Necesito volver a imputar los valores faltantes en 'Age' y 'Fare' y luego volver a escalar las características numéricas. Después de eso, definiré nuevamente las características y el objetivo.



```python
# Re-impute missing values in 'Age' and 'Fare' after dropping 'Name' and 'Ticket'
# Using the median again as it's robust to outliers.
df["Age"].fillna(df["Age"].median(), inplace=True)
df["Fare"].fillna(df["Fare"].median(), inplace=True)

print("Valores faltantes por columna después de re-imputar Age y Fare:")
print(df.isnull().sum())

# Re-scale the numerical features now that missing values are handled
numerical_features_to_scale = ['Age', 'SibSp', 'Parch', 'Fare']
scaler = StandardScaler()
df[numerical_features_to_scale] = scaler.fit_transform(df[numerical_features_to_scale])

print("\nDataFrame después de re-escalar las características numéricas:")
display(df.head())

# Define las variables predictoras (features, X) y la variable objetivo (target, y) again
# 'Survived' is the target variable.
# All other columns except 'PassengerId' (which is just an identifier) will be features.
X = df.drop(['Survived', 'PassengerId'], axis=1)
y = df['Survived']

print("\nVariables predictoras (X) después de re-procesamiento:")
display(X.head())

print("\nVariable objetivo (y) después de re-procesamiento:")
display(y.head())
```
